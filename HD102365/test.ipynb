```python
import numpy as np
import plotly.graph_objects as go

# Load data
data = np.loadtxt('/work2/lbuc/jzhao/PyORBIT_ESSP/HD102365/data/processed_data/HD102365_ESPRESSO_RV.dat')
# Assume columns: time, RV, RV_err (adapt if columns differ)
times = data[:, 0]
rv = data[:, 1]
rv_err = data[:, 2] if data.shape[1] > 2 else None

# Plot the RV time series interactively
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=times,
    y=rv,
    error_y=dict(
        type='data',
        array=rv_err if rv_err is not None else None,
        thickness=1.5,
        width=0
    ),
    mode='markers',
    marker=dict(color='royalblue', size=8, line=dict(width=1, color='black')),
    name='RV'
))

fig.update_layout(
    title="HD102365 ESPRESSO RV Time Series",
    xaxis_title="Time [BJD]",
    yaxis_title="Radial Velocity [m/s]",
    template="simple_white",
    width=900,
    height=450
)

fig.show()
```
